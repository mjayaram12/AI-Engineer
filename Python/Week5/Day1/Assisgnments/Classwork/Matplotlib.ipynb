#### Visualization chart

```python
import matplotlib.pyplot as plt
test_cases_executed = {"Passed" :  45, "Failed" : 10, "Skipped" : 5}
test_cases_executed.keys()
plt.bar(test_cases_executed.keys(), test_cases_executed.values(), color='orange',width = 0.7)
plt.title("Test Execution Results", fontdict={'size': 14, 'weight': 'bold', 'family': 'Times New Roman', 'color': 'blue'})
plt.xlabel("Test Status")
plt.ylabel("Number of Test Cases")
plt.show()
```

#### Defect Trend chart

```python
Weeks = [1, 2, 3, 4, 5, 6]
Defects = [5, 8, 6, 10, 7, 4]
plt.plot(Weeks, Defects, marker='p', linestyle='--', color='skyblue')
plt.title("Defect Trend Over Time",fontsize=14, fontweight='bold', family='serif')
plt.xlabel("Week Number",fontsize=14, fontweight='bold', family='serif')
plt.ylabel("Number of Defects",fontsize=14, fontweight='bold', family='Times New Roman')
plt.grid(True)
plt.savefig('test.png')
plt.show()
```

#### Histogram wo subplot

```python
import matplotlib.pyplot as plt
execution_times = [12, 15, 20, 18, 22, 30, 25, 16, 19, 28, 24, 14]
plt.hist(execution_times, bins=5, color='skyblue', edgecolor='black')
plt.title("Distribution of Test Execution Times")
plt.xlabel("Duration (seconds)")
plt.ylabel("Number of Test Cases")
plt.savefig("histogram._Savefig.png")
plt.show()
```

#### Histogram and Pie with sub-plot

```python
execution_times = [12, 15, 20, 18, 22, 30, 25, 16, 19, 28, 24, 14]
data = {"High" : 10, "Medium" : 15, "Low" : 5}
 
fig, axes = plt.subplots(1, 2, figsize=(18, 8))
fig.suptitle("Assignment")
 
axes[0].hist(execution_times, bins= 5, color= 'skyblue' , edgecolor = 'black' )
axes[0].set_title("Distribution of Test Execution Times")
axes[0].set_xlabel("Duration seconds")
axes[0].set_ylabel("Number of Test Cases")
 
axes[1].pie(data.values(),labels=data.keys(), autopct="%1.0f%%", startangle=90,frame=True)
axes[1].set_title("Defect Distribution by Severity")
```
